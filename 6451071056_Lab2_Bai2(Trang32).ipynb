Ô 1: Thư viện và hàm tính F-measure

```python
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix
import pandas as pd

print("Lê Anh Pha - 6451071056")

def f_measure(precision, recall):
    """F-measure = 2 * P * R / (P + R)"""
    if precision + recall == 0:
        return 0.0
    return 2 * precision * recall / (precision + recall)

def danh_gia_nhi_phan(y_true, y_pred, lop_duong):
    """Tính precision, recall, F-measure cho bài toán 2 lớp."""
    p = precision_score(y_true, y_pred, pos_label=lop_duong, zero_division=0)
    r = recall_score(y_true, y_pred, pos_label=lop_duong, zero_division=0)
    f = f_measure(p, r)
    # kiểm tra chéo với hàm của sklearn
    assert abs(f - f1_score(y_true, y_pred, pos_label=lop_duong, zero_division=0)) < 1e-9
    return p, r, f
```

Ô 2: Ví dụ ngân hàng

```python
print("Lê Anh Pha - 6451071056")

thuc_te = ['Rủi ro cao', 'Rủi ro thấp', 'Rủi ro thấp']
cart_du_doan = ['Rủi ro thấp', 'Rủi ro thấp', 'Rủi ro thấp']   # cây CART trong đề
nb_du_doan   = ['Rủi ro cao',  'Rủi ro thấp', 'Rủi ro thấp']   # Naive Bayes trong đề

ket_qua = []
for ten, du_doan in [('Cây CART', cart_du_doan), ('Naive Bayes', nb_du_doan)]:
    p, r, f = danh_gia_nhi_phan(thuc_te, du_doan, 'Rủi ro thấp')
    ket_qua.append([ten, p, r, f])

pd.DataFrame(ket_qua, columns=['Mô hình', 'Precision', 'Recall', 'F-measure'])
```

Ô 3: Thực hành 1, doanh số đồ chơi

```python
print("Lê Anh Pha - 6451071056")

thuc_te = ['Thấp', 'Thấp', 'Cao']

# >>> Điền dự đoán của 3 sản phẩm theo kết quả tính tay của bạn <<<
cay_du_doan = ['Thấp', 'Thấp', 'Thấp']     # ví dụ - thay bằng kết quả cây của bạn (câu d)
nb_du_doan  = ['Thấp', 'Thấp', 'Cao']      # ví dụ - thay bằng kết quả Naive Bayes (câu i)

ket_qua = []
for ten, du_doan in [('Cây quyết định', cay_du_doan), ('Naive Bayes', nb_du_doan)]:
    p, r, f = danh_gia_nhi_phan(thuc_te, du_doan, 'Thấp')
    ket_qua.append([ten, p, r, f])

print("Lớp dương: Thấp")
pd.DataFrame(ket_qua, columns=['Mô hình', 'Precision', 'Recall', 'F-measure'])
```

Ô 4: Thực hành 2, cảm xúc (3 lớp)

```python
print("Lê Anh Pha - 6451071056")

thuc_te = ['xấu', 'tốt', 'bình thường', 'tốt']
cac_lop = ['tốt', 'xấu', 'bình thường']

# >>> Điền dự đoán của 4 trạng thái theo kết quả của bạn <<<
cay_du_doan = ['xấu', 'tốt', 'tốt', 'tốt']          # ví dụ - thay bằng kết quả cây (câu c)
nb_du_doan  = ['xấu', 'tốt', 'bình thường', 'tốt']  # ví dụ - thay bằng kết quả Naive Bayes (câu c)

for ten, du_doan in [('Cây quyết định', cay_du_doan), ('Naive Bayes', nb_du_doan)]:
    print("==", ten, "==")
    print(confusion_matrix(thuc_te, du_doan, labels=cac_lop))   # hàng = thực tế, cột = dự đoán
    dong = []
    for lop in cac_lop:
        p = precision_score(thuc_te, du_doan, labels=[lop], average=None, zero_division=0)[0]
        r = recall_score(thuc_te, du_doan, labels=[lop], average=None, zero_division=0)[0]
        dong.append([lop, p, r, f_measure(p, r)])
    bang = pd.DataFrame(dong, columns=['Lớp', 'Precision', 'Recall', 'F-measure'])
    bang.loc[len(bang)] = ['Macro (trung bình)', bang['Precision'].mean(), bang['Recall'].mean(), bang['F-measure'].mean()]
    display(bang)
```
